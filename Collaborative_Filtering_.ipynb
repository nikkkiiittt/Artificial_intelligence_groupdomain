```python
import pandas as pd
from sklearn.metrics.pairwise import cosine_similarity
```

```python
ratings = pd.DataFrame({
    'Movie_A': [5, 5, 2, 5],
    'Movie_B': [3, 2, 5, 3],
    'Movie_C': [4, 4, 1, 5],
    'Movie_D': [0, 5, 2, 4]
}, index=['Alice', 'Bob', 'Charlie', 'David'])

print(ratings)
```

```python
similarity = cosine_similarity(ratings)

similarity_df = pd.DataFrame(
    similarity,
    index=ratings.index,
    columns=ratings.index
)

alice_sim = similarity_df['Alice']
```

```python

weighted_sum = (
    alice_sim['Bob'] * ratings.loc['Bob', 'Movie_D']
    + alice_sim['Charlie'] * ratings.loc['Charlie', 'Movie_D']
    + alice_sim['David'] * ratings.loc['David', 'Movie_D']
)

sum_similarity = (
    alice_sim['Bob']
    + alice_sim['Charlie']
    + alice_sim['David']
)

predicted_rating = weighted_sum / sum_similarity

print(round(predicted_rating, 2))

# Using cosine similarity between users, Alice's predicted rating for Movie_D (which she has not rated) was computed as 3.74, based on the ratings given by similar users Bob, Charlie and David
```
