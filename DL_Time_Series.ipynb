```python
import numpy as np

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense
```

```python
data = np.array([
    30, 31, 32, 31, 33, 34, 35,
    34, 36, 37,
    38, 37, 39, 40, 41
], dtype=float)
```

```python
X = []
y = []

for i in range(len(data) - 3):
    X.append(data[i:i+3])
    y.append(data[i+3])

X = np.array(X)
y = np.array(y)

X = X.reshape((X.shape[0], X.shape[1], 1))
```

```python
model = Sequential([
    LSTM(50, activation='relu', input_shape=(3, 1)),
    Dense(1)
])
```

```python
model.compile(
    optimizer='adam',
    loss='mse'
)

model.fit(
    X,
    y,
    epochs=100,
    verbose=0
)
```

```python
last_three = np.array([39, 40, 41]).reshape(1, 3, 1)

prediction = model.predict(last_three, verbose=0)

print("Previous 3 temperatures: 39, 40, 41")
print("Predicted next temperature:", prediction[0][0])
```
