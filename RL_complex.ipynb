```python
import numpy as np
import random
```

```python
# 4 states: 0, 1, 2, 3
# Goal = state 3

# Actions: 0 = Left, 1 = Right
q = np.zeros((4, 2))

learning_rate = 0.8
discount = 0.9
episodes = 100
```

```python
for episode in range(episodes):
    state = 0

    while state != 3:

        # Choose an action
        if random.random() < 0.2:
            action = random.randint(0, 1)
        else:
            action = np.argmax(q[state])

        # Perform action
        if action == 1:      # Right
            next_state = min(state + 1, 3)
        else:                # Left
            next_state = max(state - 1, 0)

        # Reward
        if next_state == 3:
            reward = 10
        else:
            reward = -1

        # Q-learning formula
        q[state, action] = q[state, action] + learning_rate * (
            reward
            + discount * np.max(q[next_state])
            - q[state, action]
        )

        state = next_state
```

```python
print("Learned Q-Table:")
print(q)
```

```python
state = 0
path = [state]

while state != 3:
    action = np.argmax(q[state])

    if action == 1:
        state = min(state + 1, 3)
    else:
        state = max(state - 1, 0)

    path.append(state)

print("Best path:", path)
print("Goal reached!")
```
